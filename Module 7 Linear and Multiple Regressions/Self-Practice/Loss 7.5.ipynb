```python
import seaborn as sns
import pandas as pd
import plotly.express as px
import numpy as np
```

```python
data = sns.load_dataset("tips")
data.head(5)

features = data[["total_bill"]]
tip = data["tip"]

from sklearn import linear_model
f = linear_model.LinearRegression(fit_intercept=False)
f.fit(features, tip)
```

```python
data["prediction"] = f.predict(features)
data.head(5)
```

```python
data["l2_loss"] = (data["tip"] - data["prediction"]) ** 2
data.head(5)
```

```python
np.mean(data["l2_loss"])
```

```python
data["l2_loss"].mean()
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(tip, data["prediction"])
```

```python
def mse_given_theta(theta):
    return mean_squared_error(data["total_bill"] * theta, tip)
    
```

```python
mse_given_theta(0.2)
```

```python
mse_given_theta(0.08)
```

```python
thetas = np.linspace(0.1, 0.2, 100)
thetas
```

```python
mses = [mse_given_theta(theta) for theta in thetas]
mses
```

```python
fig = px.line(x=thetas, y=mses, labels={"x": "Theta", "y": "MSE"}, title="MSE vs Theta")
fig.update_layout(xaxis_title="Theta", yaxis_title="MSE")
fig.show()  
```

```python
import scipy.optimize as opt
```

```python
opt.minimize(mse_given_theta, x0 = 0.2)
```
