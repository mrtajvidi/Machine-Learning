```python
import seaborn as sns
import pandas as pd
import plotly.express as px
import numpy as np
```

```python
data = sns.load_dataset("tips")
data.head(5)
```

```python
features = data[["total_bill"]]
tip = data["tip"]

from sklearn import linear_model
f = linear_model.LinearRegression(fit_intercept=False)
f.fit(features, tip)
f.predict([[100]])
```

```python
f.predict([[74.12]])
```

```python
data["prediction"] = f.predict(features)
```

```python
data
```

```python
import plotly.graph_objects as go
fig = go.Figure()
fig.add_trace(go.Scatter(x=data["total_bill"], y=data["tip"], mode='markers', name='Actual Tips'))
fig.add_trace(go.Scatter(x=data["total_bill"], y=data["prediction"], mode='lines', name='Predicted Tips'))
fig.update_layout(title='Total Bill vs Tip', xaxis_title='Total Bill', yaxis_title='Tip')
fig.show()
```
